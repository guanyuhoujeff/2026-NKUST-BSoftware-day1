## A5・自訂函式：把乖離率包成自己的工具

乖離率＝今天的價格離均線多遠，用百分比表示：(價格 − 均線) ÷ 均線 × 100

### A5-1・定義乖離率函式

> 這一格跑完不會有輸出，下一格才開始用它。

````python
_____ bias(price, ma):                # TODO：「定義一個新工具」要用哪個關鍵字？括號裡是它需要的兩個材料
    return (price - ma) / ma * 100    # 💡 return＝把算好的結果交回去；先算「差多少」，再除以均線
````
