## A2・list：把 5 天的收盤價排成一排

### A2-1・5 天的收盤價

> 預期：`[2385, 2400, 2425, 2415, 2370]`

````python
closes = [2385, 2400, 2425, 2415, 2370]   # 💡 中括號＝list（串列）；照時間排：10/22 在最左、10/30 在最右
closes
````

### A2-2・有幾天

## A3・for：讓電腦一天一天看過去

### A3-1・一天一天印出來

> 預期：5 行，從 `2385` 到 `2370`

````python
for c in closes:      # 💡 for＝把 closes 裡的東西一個一個拿出來，每次叫它 c
    print(c)          # 💡 往右縮排的這一行，每拿一個就做一次
````

### A3-2・手算 5 天加總

> 預期：`11995`

````python
total = 0                 # 💡 先準備一個從 0 開始的「累加器」
for c in closes:
    total = total + c     # 💡 每拿到一天，就把它加進 total
total
````

### A3-3・5 日均線

> 預期：`2399.0`

````python
ma5 = total / len(closes)   # 💡 均線＝這幾天收盤價的平均＝加總 ÷ 天數
print(ma5)                  # 💡 這就是 10/30 那天的「5 日均線」
````

### A3-4・用內建工具對答案

> 預期：`2399.0`，跟上一格一樣

````python
print(sum(closes) / len(closes))   # 💡 sum 是 Python 內建的加總；結果要跟手算的一模一樣
````

## A4・if：今天站上均線了嗎？

### A4-1・先確認今天收盤和均線

> 預期：`2370`，下一行 `2399.0`

````python
print(close)    # 💡 10/30 收盤
print(ma5)      # 💡 10/30 的 5 日均線
````

### A4-2・比大小

> 預期：`False`

````python
print(close > ma5)   # 💡 比較的結果只有兩種：True（對）或 False（錯）；2370 > 2399.0 不成立
````

### A4-3・用 if 說人話

> 預期：`在均線之下`

````python
if close > ma5:            # 💡 條件成立（True）→ 做這一行
    print("在均線之上")
else:                      # 💡 不成立（False）→ 做這一行；剛好相等也會走到這裡
    print("在均線之下")
# 💡 冒號 : 和下一行的縮排都不能少
````

### A4-4・三種情況：天花板、地板、中間

天花板和地板在 Lab 3 才會用 5 年資料算出來，這裡先用**假設值**練習判斷。`close` 自己填一個價格試試看。

> 預期：填 2370 會印出 `在地板之下`；換成 2450、2400 再跑一次，看另外兩種結果

````python
close = 2370       # 💡 自己改一個價格再跑：2450 → 天花板之上、2400 → 中間
ceiling = 2420     # 💡 假設值，不是真的天花板
floor = 2380       # 💡 假設值，不是真的地板
if close > ceiling:
    print("在天花板之上")
elif close < floor:          # 💡 elif＝否則如果：上一個條件不成立，才檢查這一個
    print("在地板之下")
else:                        # 💡 上面兩個都不成立 → 在中間
    print("在中間")
````

## A5・def：把乖離率包成自己的工具

乖離率＝今天的價格離均線多遠，用百分比表示：(價格 − 均線) ÷ 均線 × 100

### A5-1・定義乖離率函式

> 這一格跑完不會有輸出，下一格才開始用它。

````python
def bias(price, ma):                  # 💡 def＝定義一個新工具；括號裡是它需要的兩個材料
    return (price - ma) / ma * 100    # 💡 return＝把算好的結果交回去；先算「差多少」，再除以均線
````

### A5-2・拿出今天的收盤價

上一段的 `close` 可能被你改過了，這裡重新從 `closes` 拿。

> 預期：`2370`

````python
close = closes[-1]   # 💡 拿 10/30 的收盤。ma5 是 10/30 的均線，價格也要是同一天，乖離率才有意義
close
````

### A5-3・用自己的函式算乖離率

> 預期：`-1.208837015423093`

````python
bias(close, ma5)     # 💡 用的時候：把價格和均線放進去，函式把結果交回來
````

### A5-4・再加上四捨五入比較

> 預期：`-1.21`

````python
round(bias(close, ma5), 2)   # 💡 負的＝價格在均線之下；-1.21 表示比均線低 1.21%
````

### A5-5・同一個工具，換一週再用

> 預期：`1.33`

````python
week_oct = [2510, 2500, 2575, 2585, 2585]   # 💡 10/01～10/07 的收盤價
ma_oct = sum(week_oct) / len(week_oct)      # 💡 那週的 5 日均線 2551.0

round(bias(week_oct[-1], ma_oct), 2)        # 💡 正的＝在均線之上。寫一次、到處用，這就是 def 的好處
````

### ✏️ A5 小作業：寫一個漲跌幅函式

寫一個 `change_pct(today, yesterday)`，回傳漲跌幅（%）。用它算 10/30 的漲跌幅（今天 2370、昨天 2415），四捨五入到小數 2 位。
